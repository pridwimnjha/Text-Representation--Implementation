```python
from sklearn.preprocessing import OneHotEncoder

labels = ["positive", "negative", "neutral", "positive"]
encoder = OneHotEncoder()

# Reshape into a 2D array of shape (n_samples, 1)
encoded_labels = encoder.fit_transform([[label] for label in labels])

# If you want to view the result as an array
print(encoded_labels.toarray())  # Converts the sparse matrix to a dense array
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Sample text data
documents = [
    "I am happy",
    "I am sad",
    "I feel nothing",
    "I am very happy"
]

# Initialize vectorizer
vectorizer = CountVectorizer()

# Fit and transform the documents into a bag-of-words matrix
X = vectorizer.fit_transform(documents)

# Show feature names
print("Vocabulary:", vectorizer.get_feature_names_out())

# Convert the sparse matrix to an array to visualize it
print("Bag of Words Matrix:\n", X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample text data
documents = [
    "I am happy",
    "I am sad",
    "I feel nothing",
    "I am very happy"
]

# Initialize the vectorizer
vectorizer = TfidfVectorizer()

# Fit and transform the documents
X = vectorizer.fit_transform(documents)

# Show feature names (vocabulary)
print("Vocabulary:", vectorizer.get_feature_names_out())

# Convert the sparse matrix to a dense array
print("TF-IDF Matrix:\n", X.toarray())
```
